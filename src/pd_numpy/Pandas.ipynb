```python
# %pip install pandas
import pandas as pd
```

```python
df = pd.DataFrame({
    'col1': [1,2,3,4,5],
    'col2': [3,4,8,9,5],
    # 'col3': 'samir bhoyar data engineer'
    'col3': 'samir bhoyar is data engineer'.split()
})
df
```

```python
df['col4']= df['col1']**2
df
```

```python
def cond(words):
    list=[]
    for word in words:
        if word[0]=='s':
           list.append('s') 
        else: 
            list.append('z')    
    return list
df['col5'] = cond(df['col3'])
df
```

```python
df['col6']= df['col1'].apply(lambda x: 'even' if  x/2==0 else 'odd' )
df
```

```python
df['col7'] = df['col3'].apply(len)
df
```

```python
import math
df['col8'] = df['col2'].apply(math.log)
df
```
